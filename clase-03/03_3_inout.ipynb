## Entrada y salida nativa

Hasta ahora utilizamos la función `print()` para imprimir por pantalla mensajes y resultados, y esa salida puede direccionarse a un archivo utilizando el argumento opcional `file`.

Python permite además realizar lectura y escritura de información a archivos utilizando funciones incorporadas al lenguaje. Para leer o escribir un archivo primero debemos abrirlo, utilizando la función `open()`, que recibe como argumento un *string* o un objeto *Path*

````python
fname = here / 'data' / 'names.txt'     # Path()
fname1 = str(here) + '/data/names.txt'  # string
````

````python
f = open(fname)        # Abrimos el archivo para leer (Path)
f1 = open(fname1)      # Abrimos el archivo para leer (string)
````

````python
f
````

````python
f1
````

Ambos objetos se pueden usar indistintamente

````python
s = f.read()                    # Leemos el archivo
````

````python
f.close()                       # Cerramos el archivo
````

````python
print(s[:60])
````

````python
f = open(fname, encoding='utf-8')      # Abrimos el archivo (para leer)
s = f.read()
f.close()
print(s[:60])
````

Esta secuencia básica de trabajo en adecuada y muy común en el trabajo con archivos. Sin embargo, hay un potencial problema, que ocurrirá si hay algún error entre la apertura y el cierre del archivo. Para ello existe una sintaxis alternativa

````python
with open(fname) as fi:
    s = fi.read()
    print("leyendo...")
print(s[:60])
````

````python
type(fi)
````

La palabra `with` es una palabra reservada del lenguaje y la construcción se conoce como *contexto*. Básicamente dice que todo lo que está dentro del bloque se realizará en el contexto en que `fi` es el objeto de archivo abierto para lectura.

### Ejemplos

Vamos a repasar algunos de los conceptos discutidos las clases anteriores e introducir algunas nuevas funcionalidades con ejemplos

#### Ejemplo 1


````python
n = 0                           # contador
minlen = 3                      # longitud mínima
maxlen = 4                      # longitud máxima

with open(fname, 'r', encoding='utf-8') as fi:
  lines = fi.readlines()        # El resultado es una lista

print(type(lines))
print(len(lines))
````

````python
lines[:3]
````

````python
n = 0                           # contador
minlen = 3                      # longitud mínima
maxlen = 4                      # longitud máxima

with open(fname, 'r', encoding='utf-8') as fi:
  lines = fi.readlines()        # El resultado es una lista

for line in lines:
  if minlen <= len(line.strip()) <= maxlen:
    n += 1                         # Contamos la palabra
    print(line.strip(), end=', ')  # y la imprimimos sin cambio de línea

if minlen == maxlen:
  mensaje = f"Encontramos {n} palabras que tienen {minlen} letras"
else:
  mensaje = f"Encontramos {n} palabras que tienen entre {minlen} y {maxlen} letras"

print(f"\n\n{mensaje}")
````

Hemos utilizado aquí:

* Apertura, lectura, y cerrado de archivos 
* Iteración en un loop `for`
* Bloques condicionales (if/else)
* Formato de cadenas de caracteres con reemplazo
* Impresión por pantalla

 La apertura de archivos se realiza utilizando la función `open` (este es un buen momento para mirar su documentación) con dos argumentos: el primero es el nombre del archivo y el segundo el modo en que queremos abrirlo (en este caso la `r` indica lectura).

Con el archivo abierto, en la línea 9 leemos línea por línea todo el archivo. El resultado es una lista, donde cada elemento es una línea.

Recorremos la lista, y en cada elemento comparamos la longitud de la línea con ciertos valores. Imprimimos las líneas seleccionadas

Finalmente, escribimos el número total de líneas.

Veamos una leve modificación de este programa

#### Ejemplo 2

````python
"""Programa para contar e imprimir las palabras de una longitud dada"""

n = 0                           # contador
minlen = 3                      # longitud mínima
maxlen = 4                      # longitud máxima

with open(fname, 'r', encoding='utf-8') as fi:
  for line in fi:
    p = line.strip().lower()    # Desechamos espacios y cambiamos a minúsculas
    if (minlen <= len(p) <= maxlen) and (p == p[::-1]):  # ¿Que estamos filtrando acá?
      n += 1
      print(f"{n:02d}: {p}", end=', ')  # Vamos numerando las coincidencias

if minlen == maxlen:
  mensaje = f"Encontramos un total de {n} palabras capicúa que tienen {minlen} letras"
else:
  mensaje = f"Encontramos un total de {n} palabras capicúa que tienen entre {minlen} y {maxlen} letras"

print(f"\n{mensaje}")
````

Aquí en lugar de leer todas las líneas e iterar sobre las líneas resultantes, iteramos directamente sobre el archivo abierto.

Además incluimos un string al principio del archivo, que servirá de documentación, y puede accederse mediante los mecanismos usuales de ayuda de Python.

Imprimimos el número de palabra junto con la palabra, usamos `02d`, indicando que es un entero (`d`), que queremos que el campo sea de un mínimo número de caracteres de ancho (en este caso 2). Al escribirlo como `02` le pedimos que complete los vacíos con ceros.



````python
"""Programa para contar e imprimir las palabras de una longitud dada"""

n = 0                           # contador
minlen = 3                      # longitud mínima
maxlen = 5                      # longitud máxima
L = []

with open(fname, 'r') as fi:
  for line in fi:
    p = line.strip().lower()    # Desechamos espacios y cambiamos a minúsculas
    if (minlen <= len(p) <= maxlen) and (p == p[::-1]):  # Palíndromos con la longitud correcta
      n += 1
      L.append(p)  # Similar a L += [p]  -> L = L + [p]

ss = " ".join(L)   # string con las palabras separadas por un espacio

if minlen == maxlen:
  mensaje = f"Encontramos un total de {n} palabras capicúa que tienen {minlen} letras"
else:
  mensaje = f"Encontramos un total de {n} palabras capicúa que tienen entre {minlen} y {maxlen} letras"

print(f"\n{mensaje}")

with open(fname.parent / 'tmp.txt','w') as fo:
    fo.write(ss)
````


## Archivos comprimidos

Existen varias formas de reducir el tamaño de los archivos de datos.  Varios factores, tales como el sistema operativo, nuestra familiaridad con cada uno de ellos, le da una cierta preferencia a algunos de los métodos disponibles. Veamos cómo hacer para leer y escribir algunos de los formatos:


````python
import gzip
import bz2
````

````python
with gzip.open(f"{fname}.gz", 'r') as fi:
  a = fi.read()
````

````python
l= a.splitlines()
print(l[:10])
````

````python
type(l[1])
````

````python
l[0]
````

--------

**Nota:** Vemos que la función lee el archivo y devuelve los `bytes` y presenta algunos caracteres que no podemos interpretar. Por ejemplo:

```python

l[0] = "b'\\xc3\\x81frica'"
```

La notación `b"asd"` indica que la variable es del tipo "bytes", y en este caso hay un caracter que no corresponde a la codificación ascii. Para convertirlo a strings debemos decodificarlo:


```python

l[0].decode()  # o (l[0]).encode('utf-8') 
-> 'África'

```

--------


````python
l[0].decode('utf-8')
````

````python
l[0].decode()
````

````python
type(l[0].decode())
````

````python
with bz2.open(f"{fname}.bz2", 'r') as fi:
  b = fi.read()
````

````python
a == b
````

Vemos que podemos leer cualquiera de los dos formatos de una manera muy similar.
